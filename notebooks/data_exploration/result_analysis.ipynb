```python
import json
from collections import Counter
from src.solver_eq_symbolic import OP_NAME_TO_IDX

ALL_OPS = list(OP_NAME_TO_IDX.keys())  # canonical ops, index order

def count_ops(rows):
    mapping_counter = Counter()
    occurrence_counter = Counter()
    for _, row in rows.iterrows():
        ops_json = row['ops_mapping']
        if pd.isna(ops_json) if isinstance(ops_json, float) else ops_json is None:
            continue
        ops = json.loads(ops_json)  # {op_char: op_name}

        # Count per-equation occurrences from the stored equations JSON
        eq_counts = Counter()
        equations_json = row['equations']
        if equations_json and not (isinstance(equations_json, float) and pd.isna(equations_json)):
            try:
                for eq in json.loads(equations_json):
                    eq_counts[eq['lhs']['op']] += 1
            except Exception:
                pass

        for op_char, op_name in ops.items():
            mapping_counter[op_name] += 1
            occurrence_counter[op_name] += eq_counts.get(op_char, 0)

    return mapping_counter, occurrence_counter

# standard = big-endian mode; little_endian = reversed-digit mode
df_arith = df[df['solver_strategy'].isin(['standard', 'little_endian'])]
df_std   = df_arith[df_arith['solver_strategy'] == 'standard']
df_le    = df_arith[df_arith['solver_strategy'] == 'little_endian']
df_pure_concat = df[df['solver_strategy'] == 'pure_concat']

std_map,   std_occ   = count_ops(df_std)
le_map,    le_occ    = count_ops(df_le)
pure_concat_map, pure_concat_occ = count_ops(df_pure_concat)
total_map, total_occ = count_ops(df_arith)

freq = pd.DataFrame({
    'operation':        ALL_OPS,
    'std_mapping':      [std_map.get(op, 0)   for op in ALL_OPS],
    'std_occurrence':   [std_occ.get(op, 0)   for op in ALL_OPS],
    'le_mapping':       [le_map.get(op, 0)    for op in ALL_OPS],
    'le_occurrence':    [le_occ.get(op, 0)    for op in ALL_OPS],
    'pure_concat_mapping': [pure_concat_map.get(op, 0) for op in ALL_OPS],
    'pure_concat_occurrence': [pure_concat_occ.get(op, 0) for op in ALL_OPS],
    'total_mapping':    [total_map.get(op, 0) for op in ALL_OPS],
    'total_occurrence': [total_occ.get(op, 0) for op in ALL_OPS],
}).sort_values('total_mapping', ascending=False).reset_index(drop=True)

print(f"Non-concat puzzles: {len(df_arith)}  (standard={len(df_std)}, little_endian={len(df_le)}, pure_concat={len(df_pure_concat)})")
print(f"Ops with zero mapping count: {(freq['total_mapping'] == 0).sum()} / {len(ALL_OPS)}")

with pd.option_context('display.max_rows', None):
    display(freq)
```

## 7. Symbol Distribution Analysis
Analyzes how operator symbols are distributed across puzzles: frequency per symbol, per LHS position, and cross-position correlation.

```python
df['num_unique_symbols'].hist(bins=range(1, 20), edgecolor='black')
```

```python

# Operator symbols: extract position-2 (middle of 5-char LHS) from every equation in every prompt
from collections import Counter

op_sym_counts = Counter()
total_equations = 0

for _, row in df.iterrows():
    for line in row['prompt'].split('\n'):
        if ' = ' not in line or line.startswith('Now'):
            continue
        lhs = line.split(' = ', 1)[0]
        # All LHS are 5 chars; operator is always at index 2
        if len(lhs) == 5:
            op_sym_counts[lhs[2]] += 1
            total_equations += 1

sym_df = pd.DataFrame(
    [(sym, cnt, round(cnt / total_equations * 100, 2))
     for sym, cnt in op_sym_counts.most_common()],
    columns=['operator_symbol', 'equation_count', 'pct_of_equations']
)

print(f"Total equations parsed: {total_equations}")
print(f"Unique operator symbols: {len(op_sym_counts)}\n")
with pd.option_context('display.max_rows', None):
    display(sym_df)
```

```python

# Compare symbol frequency distribution across all 5 LHS positions
import numpy as np

pos_counts = [Counter() for _ in range(5)]
total = 0

for _, row in df.iterrows():
    for line in row['prompt'].split('\n'):
        if ' = ' not in line or line.startswith('Now'):
            continue
        lhs = line.split(' = ', 1)[0]
        if len(lhs) == 5:
            for i, ch in enumerate(lhs):
                pos_counts[i][ch] += 1
            total += 1

all_syms = sorted(set().union(*pos_counts))

rows = []
for i, c in enumerate(pos_counts):
    counts = [c.get(s, 0) for s in all_syms]
    top3 = ', '.join(f'{s}={n}' for s, n in c.most_common(3))
    rows.append({
        'position':    f'pos{i} (operator)' if i == 2 else f'pos{i}',
        'unique_syms': len(c),
        'min':         min(counts),
        'max':         max(counts),
        'std':         round(np.std(counts), 1),
        'top3':        top3,
    })

display(pd.DataFrame(rows))

# Pearson correlation between position frequency vectors
vecs = [np.array([c.get(s, 0) for s in all_syms], dtype=float) for c in pos_counts]
corr = np.array([[np.corrcoef(vecs[i], vecs[j])[0, 1] for j in range(5)] for i in range(5)])
corr_df = pd.DataFrame(corr,
                       index=[f'pos{i}' for i in range(5)],
                       columns=[f'pos{i}' for i in range(5)]).round(3)
print("\nPearson correlation between position frequency vectors:")
display(corr_df)
```

## 8. Operation-Symbol Mapping
Shows which operations get assigned to the three most common operator symbols (`*`, `+`, `-`) across all puzzles.

```python

# Operations mapped to *, -, + across all samples
from collections import defaultdict
target_syms = ['*', '+', '-']

# Extend the target symb with the other symbols
for sym in sym_df['operator_symbol']:
    if sym not in target_syms:
        target_syms.append(sym)

op_map = defaultdict(Counter)  # sym -> {op_name: count}

for _, row in df.iterrows():
    if pd.isna(row['ops_mapping']):
        continue
    ops = json.loads(row['ops_mapping'])
    for sym, op_name in ops.items():
        if sym in target_syms:
            op_map[sym][op_name] += 1

# Build a wide table: rows = operation names, columns = symbols
all_ops = sorted(set(op for c in op_map.values() for op in c))
table = pd.DataFrame(
    {sym: [op_map[sym].get(op, 0) for op in all_ops] for sym in target_syms},
    index=all_ops
)
table.index.name = 'operation'
table['total'] = table.sum(axis=1)
table = table.sort_values('total', ascending=False)

print("Rows = operation name, Columns = operator symbol, Values = number of puzzles mapped")
display(table)
```

```python
# Top 3 operations per symbol
top3_rows = []
for sym in target_syms:
    for op, cnt in op_map[sym].most_common(3):
        top3_rows.append({'symbol': sym, 'rank': op_map[sym].most_common().index((op, cnt)) + 1, 'operation': op, 'count': cnt})

top3_df = pd.DataFrame(top3_rows)
top3_pivot = top3_df.pivot_table(index=['rank', 'operation'], columns='symbol', values='count', fill_value=0)

# Simpler wide format: one row per (symbol, rank)
top3_wide = (
    top3_df
    .assign(label=top3_df['operation'] + ' (' + top3_df['count'].astype(str) + ')')
    .pivot(index='rank', columns='symbol', values='label')
    .fillna('')
)
top3_wide.columns.name = None
top3_wide.index.name = 'rank'

print("Top 3 operations per operator symbol (count in parentheses)")
display(top3_wide[target_syms])  # keep symbol column order
```

## 9. Samples Where Operator Appears as Last Character of RHS
Filters puzzles that contain at least one equation where the operator symbol (LHS position 2) is also the last character of the RHS — e.g. `![@({ = $$@` where `@` is both the operator and the final RHS character. Query equations are checked separately.

```python
example_prompt = """
Let's assume the following set of available operations that could govern the secret equations:

 1. mul(a, b) = a × b
 2. add(a, b) = a + b
 3. absdiff(a, b) = |a − b|  (always non-negative)
 4. sub_signed(a, b) = a − b  (result may be negative; if so, the operator symbol is attached
    to the answer as a sign)
 5. rsub_signed(a, b) = b − a  (same sign convention as sub_signed)
 6. neg_absdiff(a, b) = −|a − b|  (always non-positive; the operator symbol is always
    attached to the answer)
 7. concat_fwd(a, b) = the symbols of a followed by the symbols of b  (structural, no
    arithmetic)
 8. concat_rev(a, b) = the symbols of b followed by the symbols of a  (structural, no
    arithmetic)
 9. add_m1(a, b) = a + b − 1  (invalid if result would be negative)
10. add_p1(a, b) = a + b + 1
11. add_p2(a, b) = a + b + 2
12. mul_m1(a, b) = a × b − 1  (invalid if result would be negative)
13. mul_p1(a, b) = a × b + 1
14. absdiff_m1(a, b) = |a − b| − 1  (invalid when a = b)
15. absdiff_p1(a, b) = |a − b| + 1  (always at least 1)
16. absdiff_m2(a, b) = |a − b| − 2  (invalid when |a − b| < 2)
17. gcd(a, b) = greatest common divisor of a and b  (always non-negative)
18. lcm(a, b) = least common multiple of a and b  (0 if either operand is 0)
19. mod(a, b) = a mod b  (invalid when b = 0)
20. rmod(a, b) = b mod a  (invalid when a = 0)
21. fdiv(a, b) = ⌊a ÷ b⌋  (invalid when b = 0)
22. a2_plus_b(a, b) = a² + b

There are three possible solving strategies:

- Pure concatenation: the operator simply joins the two operands. The result is either the
  symbols of the left operand followed by those of the right operand (forward), or the symbols
  of the right operand followed by those of the left (reverse). No digit mapping is needed.

- Standard arithmetic (base 10, most-significant digit first): each unique symbol maps to a
  unique decimal digit (0–9). The left operand is decoded as L = digit(l0) × 10 + digit(l1),
  and the right operand as R = digit(r0) × 10 + digit(r1). The chosen operation is applied:
  result = OP(L, R). The magnitude is then converted to decimal digits from most to least
  significant, zero-padded on the left if the answer needs more digits, and each digit is
  mapped back to its symbol.

- Little-endian arithmetic (base 10, least-significant digit first): the same as standard
  arithmetic, except each operand's two digits are swapped before the operation — the symbol
  in the first position carries the units digit (not the tens digit), so the raw two-digit
  value X is interpreted as (X mod 10) × 10 + (X ÷ 10). The result digit sequence is
  reversed after zero-padding before mapping back to symbols.

We always use base 10.

The left-hand side of any equation has exactly 5 symbols: l0 l1 op r0 r1, where op is the
operator symbol, l0 l1 is the left operand, and r0 r1 is the right operand.

The right-hand side can have 1 to 4 symbols:
  - mag           — all symbols represent the magnitude of the result.
  - op mag        — operator symbol first, then magnitude (result is negative; sign prefix).
  - mag op        — magnitude first, operator symbol last (result is negative; sign postfix).

The goal is to find the digit mapping (symbol → digit) and operation mapping (op → function)
that satisfies the equation, then apply them to decode the query answer.

Formatted equation:
  `!*[{ = '"[`

Observations:
- The answer '"[` carries no operator symbol, so the result is non-negative and no signed
  operation is in play.
- The answer has 4 symbols, which matches the combined length of the two operands (2 + 2).
  This is consistent with either a 4-digit arithmetic result or concatenation — worth
  checking both.

First, I'll check whether this is a concatenation rule. Forward concatenation (l0 l1 r0 r1)
would give `![{. Reverse concatenation (r0 r1 l0 l1) would give [{`!. Neither matches the
answer '"[`, so concatenation is ruled out.

Next, could the operation be a negated difference? That operation always produces a
non-positive result and always attaches the operator symbol to the front of the answer. The
answer '"[` carries no such prefix, so a negated difference is impossible here.

**Finding the digit and operation mapping using standard arithmetic (base 10).**

I'll prioritise assigning digits to the symbols that appear most often — the more frequently
a symbol appears, the more it constrains the solution.

Symbol occurrence counts:
  ` : 2 occurrences  (positions l0 and rhs[3])
  [ : 2 occurrences  (positions r0 and rhs[2])
  ! : 1 occurrence   (position l1)
  { : 1 occurrence   (position r1)
  ' : 1 occurrence   (position rhs[0])
  " : 1 occurrence   (position rhs[1])

Assigning digits in that order:
  ` → 6
  [ → 9
  ! → 3
  { → 2
  ' → 5
  " → 7

Left operand:  digit(`) × 10 + digit(!) = 6 × 10 + 3 = **63**.
Right operand: digit([) × 10 + digit({) = 9 × 10 + 2 = **92**.

Applying the operation (a × b): 63 × 92 = **5796**.
The result 5796 has 4 digits and the answer has 4 positions — it fits (5796 < 10000). ✓

Converting digits back to symbols: 5 → ', 7 → ", 9 → [, 6 → ` → **'"[`**.

**Summary:**
- Digit mapping: ` → 6, ! → 3, [ → 9, { → 2, ' → 5, " → 7
- Operation: * → mul (a × b)

Note: this equation admits 20 consistent solutions. The mapping above is one valid program;
others may also satisfy the same equation.

solution: '"[`
"""

# NOTE: This sample contains around 1900 tokens
approx_token = 1900

rate = len(example_prompt) / approx_token
print(f"Approx tokens: {approx_token}  |  Prompt length: {len(example_prompt)}  |  Rate: {rate:.2f} chars/token")
```

## 13. Zero-Padding Requirement Analysis
Counts how many puzzles require **zero-padding** — where the result magnitude has fewer base-10 digits than the number of RHS magnitude positions, forcing leading zero-symbols on the left (mirrors `encode_answer` / `verify_equation` in `src/encoding.py`). Reports both the count for the final **query** (`to_predict`) equation and for **any** equation in the puzzle. Data: `data/solver_results_forcing_base_10_k_1.parquet` (823 unique puzzles, k = 1 solution each).

```python
import json
import pandas as pd
from src.encoding import _int_to_base_digits, _is_reversed_digit_mode
from src.operations import OPERATIONS
from src.op_index import CONCAT_OPS

# ── helper ────────────────────────────────────────────────────────────────────

def eq_requires_padding(eq, mapping, ops_mapping, mode, radix):
    """Return True if the equation's magnitude needs zero-padding, None if N/A.

    Mirrors the zero-pad logic in verify_equation (src/encoding.py:240-245):
    padding is needed when len(base-radix digits of |magnitude|) < len(actual_rhs).
    Returns None for concat ops, missing mappings, or invalid operations.
    """
    lhs = eq["lhs"]
    op_name = ops_mapping.get(lhs["op"])
    if op_name is None or op_name in CONCAT_OPS:
        return None                          # no arithmetic magnitude
    actual_rhs = eq["rhs"]["actual_rhs"]
    try:
        L = mapping[lhs["l0"]] * radix + mapping[lhs["l1"]]
        R = mapping[lhs["r0"]] * radix + mapping[lhs["r1"]]
    except KeyError:
        return None
    if _is_reversed_digit_mode(mode):        # little_endian: swap operand digits
        L = (L % radix) * radix + L // radix
        R = (R % radix) * radix + R // radix
    fn = OPERATIONS.get(op_name)
    if fn is None:
        return None
    raw = fn(L, R)
    if raw is None:
        return None
    mag = abs(raw)
    return len(_int_to_base_digits(mag, radix)) < len(actual_rhs)


def _loads(v):
    """Parse JSON string or pass through already-decoded value."""
    if isinstance(v, str):
        return json.loads(v) if v else {}
    return v or {}

def _load_eqs(v):
    """Parse equations column — handles both JSON string and already-decoded list."""
    if isinstance(v, list):
        return v
    if isinstance(v, str) and v:
        return json.loads(v)
    return []

# ── per-row detection ─────────────────────────────────────────────────────────
# Handles df['equations'] whether it was pre-parsed (list) or is still a JSON string.
# df['digits_mapping'] and df['ops_mapping'] are always JSON strings in this parquet.

pad_records = []
for _, row in df.iterrows():
    if pd.isna(row['solver_strategy']) or row['solver_category'] != 'arithmetic':
        continue                             # skip unsolved (19) and concat (75)

    mode      = row['solver_strategy']
    radix     = int(row['solver_radix'])
    mapping   = _loads(row['digits_mapping'])  # {symbol: digit}
    ops_map   = _loads(row['ops_mapping'])     # {op_char: op_name}
    equations = _load_eqs(row['equations'])    # list of equation dicts

    query_eq = next((eq for eq in equations if eq['type'] == 'to_predict'), None)
    if query_eq is None:
        continue

    query_pad = eq_requires_padding(query_eq, mapping, ops_map, mode, radix) is True
    any_pad   = any(eq_requires_padding(eq, mapping, ops_map, mode, radix) is True
                    for eq in equations)

    # Reconstruct a human-readable query equation string for the review table
    lhs = query_eq['lhs']
    rhs = query_eq['rhs']
    rhs_str = ''.join(rhs['actual_rhs'])
    if rhs['op'] is not None:
        rhs_str = (rhs['op'] + rhs_str) if rhs['sign_pos'] == 'prefix' else (rhs_str + rhs['op'])
    eq_str = f"{lhs['l0']}{lhs['l1']}{lhs['op']}{lhs['r0']}{lhs['r1']} = {rhs_str}"

    # Compute magnitude for the review table (only for padded rows)
    mag = None
    if query_pad:
        try:
            L = mapping[lhs['l0']] * radix + mapping[lhs['l1']]
            R = mapping[lhs['r0']] * radix + mapping[lhs['r1']]
            if _is_reversed_digit_mode(mode):
                L = (L % radix) * radix + L // radix
                R = (R % radix) * radix + R // radix
            op_name = ops_map.get(lhs['op'])
            fn = OPERATIONS.get(op_name) if op_name else None
            if fn is not None:
                raw = fn(L, R)
                mag = abs(raw) if raw is not None else None
        except Exception:
            pass

    pad_records.append({
        'id':              row['problem_id'],
        'query_eq':        eq_str,
        'magnitude':       mag,
        'rhs_width':       len(rhs['actual_rhs']),
        'solver_strategy': mode,
        'category':        row['solver_category'],
        'query_pad':       query_pad,
        'any_pad':         any_pad,
    })

pad_df = pd.DataFrame(pad_records)

# ── 1. Both counts ────────────────────────────────────────────────────────────
n_total = len(pad_df)
n_query = int(pad_df['query_pad'].sum())
n_any   = int(pad_df['any_pad'].sum())

print(f"Arithmetic puzzles analysed : {n_total}")
print()
print(f"Query (to_predict) requires zero-padding : {n_query:3d}  ({n_query / n_total * 100:.1f}%)")
print(f"Any equation requires zero-padding       : {n_any:3d}  ({n_any / n_total * 100:.1f}%)")

# ── 2. Breakdown by solver_strategy and category ──────────────────────────────
query_pad_df = pad_df[pad_df['query_pad']]

print(f"Breakdown of {n_query} query-padded puzzles by solver_strategy:")
for strat, cnt in query_pad_df['solver_strategy'].value_counts().items():
    print(f"  {strat:<15}: {cnt:3d}  ({cnt / n_query * 100:.1f}%)")

print(f"Breakdown by category:")
for cat, cnt in query_pad_df['solver_category'].value_counts(dropna=False).items():
    print(f"  {str(cat):<15}: {cnt:3d}  ({cnt / n_query * 100:.1f}%)")

# ── 3. Review table ───────────────────────────────────────────────────────────
review_df = (
    pad_df[pad_df['query_pad']][
        ['id', 'query_eq', 'magnitude', 'rhs_width', 'solver_strategy', 'category', 'any_pad']
    ]
    .reset_index(drop=True)
)
print(f"Review table — {len(review_df)} query-padded problems (magnitude digit count < rhs_width):")
with pd.option_context('display.max_rows', None, 'display.max_colwidth', None):
    display(review_df)
```

## 15. concat_fwd / concat_rev count check

Quick pre-balancing diagnostic: count how many `pure_concat` rows resolve to `concat_fwd`
vs `concat_rev`. Source rows have `ops_mapping = NaN`, so direction is inferred from the
equations (forward pattern checked first, so ambiguous rows fold into `concat_fwd`).

```python
import json
import pandas as pd
from collections import Counter

CHECK_SOURCE = "data/result_examples_1_rand_k_6.parquet"
check_df = pd.read_parquet(CHECK_SOURCE)


def concat_direction(row):
    """Return 'concat_fwd' / 'concat_rev' for a pure_concat row.

    Reuses any existing ops_mapping label; when ops_mapping is NaN, infers the
    direction from the equations (forward pattern checked first, so ambiguous
    rows fold into concat_fwd) — mirrors solver_eq_symbolic.py:_try_concat.
    """
    ops_json = row["ops_mapping"]
    if isinstance(ops_json, str):
        for name in json.loads(ops_json).values():
            if name in ("concat_fwd", "concat_rev"):
                return name

    eqs = json.loads(row["equations"])
    op = eqs[-1]["lhs"]["op"]
    relevant = [e for e in eqs if e["lhs"]["op"] == op]

    def matches(order):
        return all(
            e["rhs"]["actual_rhs"] == [e["lhs"][k] for k in order]
            for e in relevant
        )

    if matches(("l0", "l1", "r0", "r1")):  # forward first → ambiguous folds to fwd
        return "concat_fwd"
    if matches(("r0", "r1", "l0", "l1")):
        return "concat_rev"
    return None


concat_check = check_df[check_df["solver_strategy"] == "pure_concat"]
dir_counts = Counter(concat_check.apply(concat_direction, axis=1))

print(f"pure_concat rows in {CHECK_SOURCE}: {len(concat_check)}")
print(f"  concat_fwd: {dir_counts['concat_fwd']}")
print(f"  concat_rev: {dir_counts['concat_rev']}")
print(f"  unresolved: {dir_counts[None]}")
```

## 16. Balancing concat_fwd / concat_rev to 50/50

The 419 real `pure_concat` rows split unevenly: **330 forward / 81 reverse / 8 ambiguous**.  
The 8 ambiguous rows (where `l0l1 == r0r1` so forward = reverse) are folded into `concat_fwd`.

Strategy:
- **Label** every real `pure_concat` row's `ops_mapping` with `concat_fwd` or `concat_rev`.
- **Synthesize** made-up rows until both classes reach **N = 400**  
  (62 extra fwd, 319 extra rev).
- **Save** to `data/result_examples_0_result_k_6_balanced.parquet`.

```python
import json
import pandas as pd
from collections import Counter

SEED   = 42
TARGET = 400
SOURCE = "data/result_examples_1_rand_k_6.parquet"

bal_df = pd.read_parquet(SOURCE)
print(f"Loaded: {len(bal_df):,} rows, {len(bal_df.columns)} columns")

# --- Classify each pure_concat row (mirrors solver_eq_symbolic.py:178) -------
def classify_concat(row):
    """Return (label, op_glyph). Ambiguous (l0l1==r0r1) → concat_fwd."""
    lhs = json.loads(row["equations"])[-1]["lhs"]
    l0, l1, op = lhs["l0"], lhs["l1"], lhs["op"]
    r0, r1     = lhs["r0"], lhs["r1"]
    fwd = l0 + l1 + r0 + r1
    rev = r0 + r1 + l0 + l1
    ans = row["solver_answer"] if pd.notna(row["solver_answer"]) else row["answer"]
    label = "concat_fwd" if (fwd == rev or ans == fwd) else "concat_rev"
    return label, op

concat_mask = bal_df["solver_strategy"] == "pure_concat"
for idx, row in bal_df[concat_mask].iterrows():
    label, op_glyph = classify_concat(row)
    bal_df.at[idx, "ops_mapping"] = json.dumps({op_glyph: label})

# --- Report ------------------------------------------------------------------
def _label(ops_json):
    try:
        return list(json.loads(ops_json).values())[0]
    except Exception:
        return None

labeled = bal_df[concat_mask]
label_series = labeled["ops_mapping"].apply(_label)
fwd_count = (label_series == "concat_fwd").sum()
rev_count = (label_series == "concat_rev").sum()
print(f"Real pure_concat labeled:  concat_fwd={fwd_count}  concat_rev={rev_count}")
print(f"  (expected 338 fwd / 81 rev — 8 ambiguous folded into fwd)")
```

```python
import numpy as np

# --- Load reference dataset (operand + operator glyph pools) -----------------
OPERAND_SOURCE = "data/solver_results_forcing_base_10_k_1.parquet"
ref_df = pd.read_parquet(OPERAND_SOURCE)

ref_concat_mask = ref_df["solver_strategy"] == "pure_concat"

# Classify each pure_concat row in the reference dataset
fwd_glyphs: Counter = Counter()
rev_glyphs: Counter = Counter()

for _, row in ref_df[ref_concat_mask].iterrows():
    label, op_glyph = classify_concat(row)
    if label == "concat_fwd":
        fwd_glyphs[op_glyph] += 1
    else:
        rev_glyphs[op_glyph] += 1

fwd_op_symbols = list(fwd_glyphs.keys())
fwd_op_weights = np.array([fwd_glyphs[g] for g in fwd_op_symbols], dtype=float)
fwd_op_weights /= fwd_op_weights.sum()

rev_op_symbols = list(rev_glyphs.keys())
rev_op_weights = np.array([rev_glyphs[g] for g in rev_op_symbols], dtype=float)
rev_op_weights /= rev_op_weights.sum()

print(f"Operator glyph pools from {OPERAND_SOURCE}:")
print("  concat_fwd glyphs:", dict(fwd_glyphs.most_common()))
print("  concat_rev glyphs:", dict(rev_glyphs.most_common()))

# --- Operand glyph pool from the reference dataset ---------------------------
operand_pool: Counter = Counter()
for v in ref_df["equations"]:
    try:
        for eq in json.loads(v):
            for k in ("l0", "l1", "r0", "r1"):
                operand_pool[eq["lhs"][k]] += 1
    except Exception:
        pass

operand_glyphs  = list(operand_pool.keys())
operand_weights = np.array([operand_pool[g] for g in operand_glyphs], dtype=float)
operand_weights /= operand_weights.sum()
print(f"  Operand pool: {len(operand_glyphs)} distinct glyphs")

# --- Real LHS set to avoid collisions (from the dataset being balanced) ------
real_lhs_set: set = set()
for _, row in bal_df[concat_mask].iterrows():
    lhs = json.loads(row["equations"])[0]["lhs"]
    real_lhs_set.add((lhs["l0"], lhs["l1"], lhs["op"], lhs["r0"], lhs["r1"]))
print(f"Real concat LHSes to avoid: {len(real_lhs_set)}")
```

```python
PROMPT_PREFIX = (
    "In Alice's Wonderland, a secret set of transformation rules is applied "
    "to equations. Below are a few examples:\n\nNow, determine the result for: "
)

def make_equations_json(l0, l1, op, r0, r1, answer_syms):
    eq = {
        "type": "to_predict",
        "lhs": {"l0": l0, "l1": l1, "op": op, "r0": r0, "r1": r1},
        "rhs": {"op": None, "sign_pos": None, "actual_rhs": answer_syms},
    }
    return json.dumps([eq], ensure_ascii=False)


def synthesize_concat(direction: str, n: int, rng) -> pd.DataFrame:
    assert direction in ("fwd", "rev")
    op_syms = fwd_op_symbols if direction == "fwd" else rev_op_symbols
    op_w    = fwd_op_weights if direction == "fwd" else rev_op_weights

    rows, seen = [], set()
    attempts, max_attempts = 0, n * 50

    while len(rows) < n and attempts < max_attempts:
        attempts += 1
        op = str(rng.choice(op_syms, p=op_w))
        l0, l1, r0, r1 = [str(g) for g in rng.choice(operand_glyphs, size=4, p=operand_weights)]
        if (l0, l1) == (r0, r1):
            continue
        key = (l0, l1, op, r0, r1)
        if key in seen or key in real_lhs_set:
            continue
        seen.add(key)

        if direction == "fwd":
            answer_syms, answer, op_name = [l0, l1, r0, r1], l0+l1+r0+r1, "concat_fwd"
        else:
            answer_syms, answer, op_name = [r0, r1, l0, l1], r0+r1+l0+l1, "concat_rev"

        rows.append({
            "problem_id":                   f"synth_concat_{direction}_{len(rows)}",
            "prompt":                       PROMPT_PREFIX + l0+l1+op+r0+r1,
            "answer":                       answer,
            "solver_answer":                answer,
            "solver_correct":               True,
            "solver_strategy":              "pure_concat",
            "solver_category":              "concatenation",
            "problem_category":              None,
            "ops_mapping":                  json.dumps({op: op_name}, ensure_ascii=False),
            "digits_mapping":               None,
            "solver_tier":                  None,
            "solver_radix":                 None,
            "ops_candidates":               None,
            "equations":                    make_equations_json(l0, l1, op, r0, r1, answer_syms),
            "solution_count":               1,
            "standard_solution_count":      0,
            "little_endian_solution_count": 0,
            
        })

    if len(rows) < n:
        raise RuntimeError(f"Only synthesized {len(rows)}/{n} {direction} rows after {attempts} attempts")
    return pd.DataFrame(rows)


rng = np.random.default_rng(SEED)

n_fwd_needed = TARGET - fwd_count   # 400 - 338 = 62
n_rev_needed = TARGET - rev_count   # 400 -  81 = 319
print(f"Need to synthesize: {n_fwd_needed} concat_fwd, {n_rev_needed} concat_rev")

synth_fwd = synthesize_concat("fwd", n_fwd_needed, rng)
print(f"  ✓ concat_fwd: {len(synth_fwd)} rows generated")
synth_rev = synthesize_concat("rev", n_rev_needed, rng)
print(f"  ✓ concat_rev: {len(synth_rev)} rows generated")
```

```python
col_order = list(bal_df.columns)
balanced  = pd.concat([bal_df, synth_fwd, synth_rev], ignore_index=True)[col_order]

# --- Verification ------------------------------------------------------------
def _label(ops_json):
    try:
        return list(json.loads(ops_json).values())[0]
    except Exception:
        return None

bal_concat   = balanced[balanced["solver_strategy"] == "pure_concat"]
label_counts = bal_concat["ops_mapping"].apply(_label).value_counts()
print("Concat label counts after balancing:")
print(label_counts.to_string())

assert label_counts.get("concat_fwd", 0) == TARGET, f"concat_fwd={label_counts.get('concat_fwd')}, expected {TARGET}"
assert label_counts.get("concat_rev", 0) == TARGET, f"concat_rev={label_counts.get('concat_rev')}, expected {TARGET}"

synth = balanced[balanced["id"].astype(str).str.startswith("synth_")]
assert (synth["answer"].str.len() == 4).all(),           "Synthetic answer not length 4"
assert (synth["answer"] == synth["solver_answer"]).all(), "answer != solver_answer in synthetic"
assert list(balanced.columns) == col_order,               "Column order mismatch"

OUT = "data/result_examples_1_result_k_6_balanced.parquet"
balanced.to_parquet(OUT, index=False)

n_real = len(balanced) - len(synth)
print(f"\n✓ All assertions passed")
print(f"  Total rows  : {len(balanced):,}  (real {n_real:,} + synthetic {len(synth):,})")
print(f"  concat_fwd  : {label_counts['concat_fwd']}")
print(f"  concat_rev  : {label_counts['concat_rev']}")
print(f"  Written to  : {OUT}")
```
